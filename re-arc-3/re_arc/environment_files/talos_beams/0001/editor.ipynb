# Talos Beams — level editor

Lay out levels visually; the connections stay as small JSON fields you edit by hand.

1. Run the **setup** cell once, then run the **editor** cell.
2. Pick a **brush** and click tiles to paint the layout (walls, doors `1–9`, sources `r/g/b`, targets `R/G/B`, connectors `o`, player `@`, exit `X`).
3. The **preview** on the right is rendered by the real game.
4. **Connections are manual JSON:** `doors` maps a door digit to the matching target colour (e.g. `{"1": "R"}`); `edges` are initial wires as pairs of tile coords (e.g. `[[[2,2],[4,4]]]`).
5. **Save levels.json** writes the current level back (matched by name). The game loads that file.
